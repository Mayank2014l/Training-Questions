```python
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.read_csv("97.loan_approval_dataset_200.csv")
df.head()
```

```python
X = df.drop("LoanStatus", axis=1)

y = df["LoanStatus"].map({
    "Rejected": 0,
    "Approved": 1
})

X.head()
```

```python
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

model
```

```python
model.fit(X, y)
```

```python
applicant = [[
    80000,   # ApplicantIncome
    30000,   # CoapplicantIncome
    250000,  # LoanAmount
    240,     # LoanTermMonths
    750,     # CreditScore
    8,       # EmploymentYears
    1,       # ExistingLoans
    2,       # Dependents
    2        # PropertyAreaCode
]]

prediction = model.predict(applicant)
```

```python
if prediction[0] == 1:
    print("Predicted Loan Status is: Approved")
else:
    print("Predicted Loan Status is: Rejected")
```

```python
plt.figure(figsize=(20, 10))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Rejected", "Approved"],
    filled=True
)

plt.show()
```
