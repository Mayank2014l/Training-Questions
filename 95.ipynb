```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
import matplotlib.pyplot as plt
```

```python
X = [[1],[2],[3],[4],[5],[6],[7],[8],[9]]  # study hours
y = [0,0,0,0,1,1,1,1,1] #fail/pass
```

```python
model = DecisionTreeClassifier() # max_depth=3,  random_state=42
model
```

```python
model.fit(X,y)
```

```python
hours = 4.6
prediction = model.predict([[hours]])
```

```python
if prediction[0]==1:
    print("Predicted Result is: Pass")
else:
    print("Predicted Result is: Fail")
```

```python
plot_tree(model,feature_names=['study-hours'],class_names=["Fail","Pass"],filled=True)
plt.show
# gini means 0,0 means No Mixed Value
# gini means 0.something Mixed Value 0,1
```
