```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")
print("done")
```

```python
df['class']=dataset['target']
df
```

```python
X=df[['sepal length (cm)', 'sepal width (cm)', 'petal length (cm)','petal width (cm)']]
y=df['class']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state = 43)
```

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
model = LogisticRegression()
model
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

new_y_pred = np.round(y_pred, 2)
result = pd.DataFrame({'Actual Flower': y_test.values, 'Predicted Flower': new_y_pred})
result.head(10)
```

```python
name={0:"Satosa",1:"Versicolor",2:"Verginica"}
spl = (input("Enter sepal length: "))
spw = int(input("Enter sepal width: "))
ptl = int(input("Enter petal length: "))
ptw = int(input("Enter petal width: "))

new_data = pd.DataFrame(
    [[spl,spw,ptl,ptw]],
)

predicted = model.predict(new_data)[0]
print("Model Predicted for New data",predicted)
print("Flower Name is", name[predicted])
print("Another Ways")
print("Flower name is",dataset.target_names[predicted])
```
