```python
import numpy as np
import pandas as pd
```

```python
data = {
    "age": [20,21,np.nan,22,23],
    "marks": [80,90,np.nan,75,np.nan]
}
df = pd.DataFrame(data)
df
```

```python
df["age"]=df["age"].mean()
df
```

```python
df["age"]=df["age"].median()
df
```

```python
df["marks"]=df["marks"].median()
df
```
