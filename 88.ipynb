```python
missing_value_per = df.isnull().sum()/df.shape[0]*100
missing_value_per
```

```python
m_value = missing_value_per[missing_value_per > 20].keys()
m_value
```

```python
drop_col = df.drop(columns = m_value)
drop_col.shape
```

```python
# Accessing Nuemrical
df3 = drop_col.select_dtypes(include=['int64','float64'])
df3.head()
```

```python
plt.figure(figsize=(16, 9))
sns.heatmap(df3.isnull())
```

```python
df3[df3.isnull().any(axis=1)]
```

```python
df3.isnull().sum()
```

```python
missing_num_var = [var for var in df3.columns if df3[var].isnull().sum() > 0]

missing_num_var
```

```python
df4=df3.fillna(df3.mean())
df4.isnull().sum().sum()
```

```python
df5 = df3.fillna(df3.mean())
df5.isnull().sum().sum()
```

```python
df_concat = pd.concat([df3[missing_num_var],df4[missing_num_var],df5[missing_num_var]],axis = 1)
df_concat
```

```python
plt.figure(figsize=(16,9))
sns.heatmap(df_concat.isnull())
plt.show()
```
